## BRILLO 


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def ajustar_brillo_eficiente(image, valor=0):
    return cv2.convertScaleAbs(image, alpha=1.0, beta=valor)


img = cv2.imread('cabina_avior.jpg')


bril = -65 
img_fin = ajustar_brillo_eficiente(img, bril)


plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagen Base (Cabina Avior)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_fin, cv2.COLOR_BGR2RGB))
plt.title(f'Brillo Aumentado (+{bril})')
plt.axis('off')

plt.tight_layout()
plt.show()
```
