## MOIRÉ


El efecto muaré o moiré es una percepción visual que se produce al ver un conjunto de líneas o puntos que se superponen a otro conjunto de líneas o puntos, donde los conjuntos difieren en tamaño relativo, ángulo o espaciado. 

El término tiene su origen en el muaré (moiré, en francés), un tejido realizado tradicionalmente en seda, aunque actualmente también se puede encontrar de algodón o de fibras sintéticas. Su principal característica es que presenta un aspecto ondulado. El muaré se obtiene presionando dos capas de tejido cuando éste está húmedo. El espaciado similar pero simétrico no simétrico entre sus  hilos crea un patrón característico que permanece después del secado de la tela.

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from PIL import Image


img_original = np.array(Image.open('imagen2.jpg').convert('RGB'))


img_suave = cv2.GaussianBlur(img_original, (15, 15), 0)


desenfoque = cv2.medianBlur(img_original, 7)


limpia = cv2.bilateralFilter(desenfoque, 9, 75, 75)


gauss_map = cv2.GaussianBlur(limpia, (5, 5), 1.0)

img_nitida = cv2.addWeighted(limpia, 2.0, gauss_map, -1.0, 0)


plt.figure(figsize=(18, 6))


plt.subplot(1, 3, 1)
plt.imshow(img_original)
plt.title('Original Con Moiré')
plt.axis('off')


plt.subplot(1, 3, 2)
plt.imshow(img_suave)
plt.title('Suavizado Simple')
plt.axis('off')


plt.subplot(1, 3, 3)
plt.imshow(img_nitida)
plt.title('Limpia Y Nitidez Reforzada')
plt.axis('off')

plt.suptitle('Evolución de Limpieza y Nitidez', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
