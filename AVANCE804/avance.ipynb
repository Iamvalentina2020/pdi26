## CONTRASTE

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def constraste(img, factor=1.0):
    img=img.astype(np.float32)
    media= np.mean(img)
    img_contraste= media +(img - media) * factor
    img_contraste= np.clip(img_contraste, 0, 255).astype(np.uint8)
    return img_contraste
img= cv2.imread('cabina_avior.jpg', 0)
contrast=9
img_fin= constraste(img, contrast)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.title('Original')
plt.subplot(1,2,2)
plt.imshow(img_fin,cmap='gray')
plt.title('Contrast Image')
plt.show()
```

## A COLORES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def constraste(img, factor=1.0):
    img=img.astype(np.float32)
    for i in range(3):
        media= np.mean(img[:,:,i])
        img[:,:,i]= media+(img[:,:,i] - media) * factor
        img[:,:,i]= np.clip(img[:,:,i], 0, 255)
    
    return img.astype(np.uint8)
img= cv2.imread('cabina_avior.jpg')
contrast=3
img_fin= constraste(img, contrast)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(img_fin, cv2.COLOR_BGR2RGB))
plt.title('contraste')
plt.show()
```
