## EJERCICIO 1

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def recortar_roi(imagen, y1, y2, x1, x2):
    
    recorte = imagen[y1:y2, x1:x2]
    return recorte

imagen_p = cv2.imread('imagenP.jpg')
img_rgb = cv2.cvtColor(imagen_p, cv2.COLOR_BGR2RGB)


punto_interes = recortar_roi(img_rgb, 150, 420, 360, 480)

plt.figure(figsize=(15,7))
plt.subplot(1,2,1)
plt.title('Imagen Original')
plt.imshow(img_rgb)
plt.subplot(1,2,2)
plt.title('Punto de Interés (Recorte)')
plt.imshow(punto_interes)
plt.show()
```

## EJERCICIO 2

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def generar_degradado():
    v = np.linspace(0, 255, 256).astype(np.uint8)
    degradado = np.tile(v, (256, 1)).T
    return cv2.applyColorMap(degradado, cv2.COLORMAP_VIRIDIS)

def posterizar(imagen, niveles):
    factor = 256 // niveles
    return (imagen // factor) * factor

def solarizar(imagen, umbral):
    return np.where(imagen < umbral, imagen, 255 - imagen)

img_deg = generar_degradado()
img_deg_rgb = cv2.cvtColor(img_deg, cv2.COLOR_BGR2RGB)
img_post = posterizar(img_deg_rgb, 4)
img_sol = solarizar(img_deg_rgb, 127)

plt.figure(figsize=(15,5))
plt.subplot(1,3,1); plt.title('Degradado Lineal'); plt.imshow(img_deg_rgb)
plt.subplot(1,3,2); plt.title('Posterización (4 niveles)'); plt.imshow(img_post)
plt.subplot(1,3,3); plt.title('Solarización'); plt.imshow(img_sol)
plt.show()
```

## EJERCICIO 3

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def crear_tablero(n=8, tam=100):

    blanco = np.ones((tam, tam), dtype=np.uint8) * 255
    negro = np.zeros((tam, tam), dtype=np.uint8)
    

    f1 = np.hstack([blanco, negro] * (n//2))
    f2 = np.hstack([negro, blanco] * (n//2))
    

    tablero = np.vstack([f1, f2] * (n//2))
    return tablero

ajedrez = crear_tablero()
plt.figure(figsize=(7,7))
plt.title('Tablero de Ajedrez (100x100 px por casilla)')
plt.imshow(ajedrez, cmap='gray')
plt.show()
```

## EJERCICIO 4

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def generar_degradado_base():
    v = np.linspace(0, 255, 256).astype(np.uint8)
    degradado = np.tile(v, (256, 1)).T
    return cv2.applyColorMap(degradado, cv2.COLORMAP_VIRIDIS)

def posterizar(imagen, niveles):
    factor = 256 // niveles
    return (imagen // factor) * factor

def solarizar(imagen, umbral):

    return np.where(imagen < umbral, imagen, 255 - imagen)


img_base = generar_degradado_base()
img_base_rgb = cv2.cvtColor(img_base, cv2.COLOR_BGR2RGB)

img_post = posterizar(img_base_rgb, 4)
img_sol = solarizar(img_base_rgb, 128)

plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.title('Posterización (4 niveles)')
plt.imshow(img_post)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Solarización (Umbral 128)')
plt.imshow(img_sol)
plt.axis('off')
plt.show()

paletas_ids = [
    cv2.COLORMAP_AUTUMN, cv2.COLORMAP_BONE, cv2.COLORMAP_JET, 
    cv2.COLORMAP_WINTER, cv2.COLORMAP_RAINBOW, cv2.COLORMAP_OCEAN, 
    cv2.COLORMAP_SUMMER, cv2.COLORMAP_SPRING, cv2.COLORMAP_COOL, 
    cv2.COLORMAP_HSV, cv2.COLORMAP_PINK, cv2.COLORMAP_HOT
]

nombres_paletas = [
    "Autumn", "Bone", "Jet", "Winter", "Rainbow", "Ocean", 
    "Summer", "Spring", "Cool", "HSV", "Pink", "Hot"
]

plt.figure(figsize=(16, 12))
for i in range(len(paletas_ids)):
    v = np.linspace(0, 255, 256).astype(np.uint8)
    gray_deg = np.tile(v, (256, 1)).T
    
    res_bgr = cv2.applyColorMap(gray_deg, paletas_ids[i])
    res_rgb = cv2.cvtColor(res_bgr, cv2.COLOR_BGR2RGB)
    
    plt.subplot(3, 4, i + 1)
    plt.title(f'Paleta: {nombres_paletas[i]}')
    plt.imshow(res_rgb)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

## EJERCICIO 5

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def aplicar_degradado_sobre_imagen(imagen):
    h, w = imagen.shape
    gradiente = np.linspace(1, 0, h).reshape(h, 1)
    mascara = np.tile(gradiente, (1, w))
    return (imagen * mascara).astype(np.uint8)

img1 = cv2.imread('imagen1.jpg', 0)
img_final = aplicar_degradado_sobre_imagen(img1)

plt.figure(figsize=(15,7))
plt.subplot(1,2,1); plt.title('Imagen Original'); plt.imshow(img1, cmap='gray')
plt.subplot(1,2,2); plt.title('Imagen con Degradado'); plt.imshow(img_final, cmap='gray')
plt.show()
```

## EJERCICIO 6

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def crear_imagen_cuadrados():
    img = np.zeros((400, 400, 3), dtype=np.uint8)
    cv2.rectangle(img, (50, 50), (150, 150), (255, 0, 0), -1)   
    cv2.rectangle(img, (200, 200), (350, 350), (0, 255, 0), -1) 
    cv2.rectangle(img, (100, 250), (180, 330), (0, 0, 255), -1) 
    return img

img_original = crear_imagen_cuadrados()
img_gray = cv2.cvtColor(img_original, cv2.COLOR_RGB2GRAY)

bayer_bg = cv2.cvtColor(img_gray, cv2.COLOR_BayerBG2RGB)
bayer_rg = cv2.cvtColor(img_gray, cv2.COLOR_BayerRG2RGB)

plt.figure(figsize=(15,5))
plt.subplot(1,3,1); plt.title('Original (Cuadrados)'); plt.imshow(img_original)
plt.subplot(1,3,2); plt.title('Filtro Bayer BG'); plt.imshow(bayer_bg)
plt.subplot(1,3,3); plt.title('Filtro Bayer RG'); plt.imshow(bayer_rg)
plt.show()
```
