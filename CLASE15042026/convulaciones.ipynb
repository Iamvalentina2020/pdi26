```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread('imagenP.jpg', cv2.IMREAD_GRAYSCALE)


kernel = np.ones((3, 3), np.float32) / 9
suavizada = cv2.filter2D(img, -1, kernel)


kernel2 = np.ones((7, 7), np.float32) / 49
suavizada2 = cv2.filter2D(img, -1, kernel2)


plt.subplot(131)
plt.imshow(img, cmap='gray')
plt.title('Imagen original')
plt.axis('off')

plt.subplot(132)
plt.imshow(suavizada, cmap='gray')
plt.title('Suavizado 3x3')
plt.axis('off')

plt.subplot(133)
plt.imshow(suavizada2, cmap='gray')
plt.title('Suavizado 7x7')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## PRACTICA

```python

import numpy as np
import matplotlib.pyplot as plt
import cv2


img_color = cv2.imread('imagenP.jpg')


img_color = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)


img_gray = cv2.cvtColor(img_color, cv2.COLOR_RGB2GRAY)


kernels = [
    ("desenfoque", np.array([(1,1,1),(1,1,1),(1,1,1)], dtype=np.float32)/9.0),

    ("filtro de tipo gauss", np.array([(1,2,3,1,1),
                                       (2,7,11,7,2),
                                       (3,11,17,11,3),
                                       (2,7,11,7,2),
                                       (1,2,3,2,1)], dtype=np.float32)/119.0),

    ("deteccion de bordes", np.array([(0,1,0),(1,-4,1),(0,1,0)], dtype=np.float32)),

    ("filtro de tipo sobel", np.array([(-1,0,1),(-2,0,2),(-1,0,1)], dtype=np.float32)),

    ("realce de bordes", np.array([(0,0,0),(-1,1,0),(0,0,0)], dtype=np.float32)),

    ("enfoque", np.array([(0,-1,0),(-1,5,-1),(0,-1,0)], dtype=np.float32)),

    ("filtro de tipo sharpen", np.array([(1,-2,1),(-2,5,-2),(1,-2,1)], dtype=np.float32)),

    ("filtro norte", np.array([(1,1,1),(1,-2,1),(-1,-1,-1)], dtype=np.float32)),

    ("filtro este", np.array([(-1,1,1),(-1,-2,1),(-1,1,1)], dtype=np.float32)),

    ("repujado", np.array([(-2,-1,0),(-1,1,1),(0,1,2)], dtype=np.float32)),
]

def mostrar_resultados(imagen, es_gris=False):
    plt.figure(figsize=(18,6))

    # Imagen original
    plt.subplot(2,6,1)
    if es_gris:
        plt.imshow(imagen, cmap='gray')
    else:
        plt.imshow(imagen)
    plt.title('Original')
    plt.axis('off')

   
    for i, (nombre, kernel) in enumerate(kernels, start=2):
        salida = cv2.filter2D(imagen, -1, kernel)

        plt.subplot(2,6,i)
        if es_gris:
            plt.imshow(salida, cmap='gray')
        else:
            plt.imshow(salida)

        plt.title(nombre)
        plt.axis('off')

    plt.tight_layout()
    plt.show()

mostrar_resultados(img_gray, es_gris=True)
mostrar_resultados(img_color, es_gris=False)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

def convolucion_media_h(imagen, tkernel):
    kernel=np.ones(tkernel, np.float32)/np.prod(tkernel)
    imagen_c=cv2.filter2D(imagen, -1, kernel)
    return imagen_c

img= cv2.imread('imagen1.jpg', 0)  

img_fin= convolucion_media_h(img, 15)
plt.figure(figsize=(10,5))
plt.subplot(121)

plt.imshow(img,cmap='gray')
plt.subplot(122)
plt.imshow(img_fin,cmap='gray')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

def convolucion_media_h(imagen, tkernel):
    kernel = np.ones((tkernel, tkernel), np.float32) / (tkernel * tkernel)
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

img = cv2.imread('imagen1.jpg')


img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


img_fin = convolucion_media_h(img_rgb, 15)


plt.figure(figsize=(10,5))

plt.subplot(121)
plt.title("Original")
plt.imshow(img_rgb)

plt.subplot(122)
plt.title("Filtrada")
plt.imshow(img_fin)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

random_img= np.random.randint(0, 256, size=(256,256), dtype=np.uint8)
img_gauss=cv2.GaussianBlur(random_img,(5,5),0)

plt.figure(figsize=(10,5))

plt.subplot(121)
plt.title("Original")
plt.imshow(random_img,cmap='gray')

plt.subplot(122)
plt.title("Filtrada")
plt.imshow(img_gauss,cmap='gray')

plt.show()
```
