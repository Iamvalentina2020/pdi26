## CONVULACIONES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread('imagenP.jpg', cv2.IMREAD_GRAYSCALE)


kernel = np.ones((3, 3), np.float32) / 9
suavizada = cv2.filter2D(img, -1, kernel)


kernel2 = np.ones((7, 7), np.float32) / 49
suavizada2 = cv2.filter2D(img, -1, kernel2)


plt.subplot(131)
plt.imshow(img, cmap='gray')
plt.title('Imagen original')
plt.axis('off')

plt.subplot(132)
plt.imshow(suavizada, cmap='gray')
plt.title('Suavizado 3x3')
plt.axis('off')

plt.subplot(133)
plt.imshow(suavizada2, cmap='gray')
plt.title('Suavizado 7x7')
plt.axis('off')

plt.tight_layout()
plt.show()
```
