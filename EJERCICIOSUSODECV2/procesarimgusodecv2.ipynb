```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def procesar_y_mostrar(ruta_imagen, nombre_guardado):
    # 1. Cargar imagen en BGR
    img_bgr = cv2.imread(ruta_imagen)
    assert img_bgr is not None, f"Error: No se encontró la imagen {ruta_imagen}"


    # 2. Convertir a RGB (para mostrar en Matplotlib)
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)


    # 3. Convertir a Escala de Grises
    img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)


    # 4. Guardar imágenes en tu ordenador (BGR y Grises)
    cv2.imwrite(f"{nombre_guardado}_bgr.jpg", img_bgr)
    cv2.imwrite(f"{nombre_guardado}_gris.jpg", img_gray)


    # 5. Separar los canales de la imagen a color (Recordando que cv2 lee en BGR)
    b = img_bgr[:,:,0]
    g = img_bgr[:,:,1]
    r = img_bgr[:,:,2]

    
    plt.figure(figsize=(15, 10))


    # Modo de subplot 1
    # Original en formato de color
    plt.subplot(231) 
    plt.title('1. Original en Color (RGB)')
    plt.imshow(img_rgb)
    plt.axis('off')


    # Modo de subplot 2
    # Imagen en escala de grises 
    plt.subplot(2, 3, 2) 
    plt.title('2. Escala de Grises')
    plt.imshow(img_gray, cmap='gray')
    plt.axis('off')


    # Color Falso (usando un cmap sobre la de grises)
    plt.subplot(2, 3, 3)
    plt.title('3. Falso Color (jet)')
    plt.imshow(img_gray, cmap='jet')
    plt.axis('off')


    # Canales separados en escala de grises
    plt.subplot(2, 3, 4)
    plt.title('4. Canal Rojo')
    plt.imshow(r, cmap='gray')
    plt.axis('off')
    

    plt.subplot(2, 3, 5)
    plt.title('5. Canal Verde')
    plt.imshow(g, cmap='gray')
    plt.axis('off')

    plt.subplot(2, 3, 6)
    plt.title('6. Canal Azul')
    plt.imshow(b, cmap='gray')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
# Segmento código para la PRIMERA imagen
procesar_y_mostrar('foto1.jpg', 'mi_foto_1')
```

```python
# Segmento código para la SEGUNDA imagen
procesar_y_mostrar('foto2.jpg', 'mi_foto_2')
```

```python
# Segmento código para la TERCERA imagen
procesar_y_mostrar('foto3.jpg', 'mi_foto_3')
```
